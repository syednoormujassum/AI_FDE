## LAMBDA Func

syntax:

lambda input variable : operation to be performed.

```python
mul=lambda x,y,z:x*y*z

print(mul(1,3,7))


```

```python
even= lambda y:y%2==0

if even(10):
    print("even")
else:
    print("odd")







```

# List comprehension

expression followed by for condition followed by zero or more for or if conditions

```python
lst3=random.sample(range(1,51),10)
```

```python
lst3
```

```python
lstt1=[]
def square(lstt):
    for i in lstt:
        lstt1.append(i*i)
    return lstt1
```

```python
square(lst3)
```

same using list comprehension

```python
[i*i for i in lst3]
```

taking only even numbers in list lst3

```python
[i*i for i in lst3 if i%2==0]
```

```python
[print(f"square root of {i} is {i*i}") for i in lst3 if i%2==0]
```

```python
[(f"square root of {i} --->  {i*i}") for i in lst3 if i%2==0]
```
