# Closures

function inside a function is called closure

```python
def Welcome(name):
    def sub_welcome():
        print(f"Welcome {name}")
        print("Come Again Later")
    return sub_welcome()
```

```python
Welcome("syed")
```

```python
def Welcome2():
    name="syed"
    def sub_welcome():
        print(f"Welcome {name}")
        print("Come Again Later")
    return sub_welcome()
```

```python
Welcome2()
```

# Decorators

decorators are also closures but they use function as a parameter which can be used inside another function

```python
def greeting(func):
    def email_greet():
        print(f"Welcome Syed ")
        print("thanks for using our service")
        func()
        print("Come again soon!!!!")
    return email_greet()
```

```python
@greeting
def gym():
    print("Did you enjoy your workout!!")
```

```python
@greeting
def stay():
    print("Hope Stay was good..")
```

2nd method to call decorators

```python
def stay():
    print("Hope Stay was good..")
```

```python
greeting(stay)
```

```python
greeting(gym)
```
